```python
import os
import shutil

dataset_root = "/content/drive/MyDrive/newMaskedDataset/newMaskedDataset"
output_root = "/content/coffee_classifier"

class_names = {
    0: "cut",
    1: "good",
    2: "husk",
    3: "immature",
    4: "parchment",
    5: "partial-black",
    6: "shell"
}

for split in ["train", "valid", "test"]:

    image_dir = os.path.join(dataset_root, split, "images")
    label_dir = os.path.join(dataset_root, split, "labels")

    for label_file in os.listdir(label_dir):

        if not label_file.endswith(".txt"):
            continue

        with open(os.path.join(label_dir, label_file)) as f:
            lines = [line.strip() for line in f if line.strip()]

        # Skip empty labels
        if len(lines) == 0:
            continue

        class_id = int(lines[0].split()[0])
        class_name = class_names[class_id]

        image_file = label_file.replace(".txt", ".jpg")

        src = os.path.join(image_dir, image_file)

        if not os.path.exists(src):
            continue

        dst_dir = os.path.join(output_root, split, class_name)
        os.makedirs(dst_dir, exist_ok=True)

        shutil.copy(src, os.path.join(dst_dir, image_file))

print("Done!")
```

```python
import torch
from torchvision import datasets, transforms
from torch.utils.data import DataLoader

train_transform = transforms.Compose([
    transforms.Resize((224,224)),
    transforms.RandomHorizontalFlip(),
    transforms.RandomRotation(10),
    transforms.ColorJitter(
        brightness=0.2,
        contrast=0.2
    ),
    transforms.ToTensor()
])

val_transform = transforms.Compose([
    transforms.Resize((224,224)),
    transforms.ToTensor()
])

train_ds = datasets.ImageFolder(
    "/content/coffee_classifier/train",
    transform=train_transform
)

val_ds = datasets.ImageFolder(
    "/content/coffee_classifier/valid",
    transform=val_transform
)

test_ds = datasets.ImageFolder(
    "/content/coffee_classifier/test",
    transform=val_transform
)

train_loader = DataLoader(
    train_ds,
    batch_size=32,
    shuffle=True,
    num_workers=2
)

val_loader = DataLoader(
    val_ds,
    batch_size=32,
    shuffle=False,
    num_workers=2
)

test_loader = DataLoader(
    test_ds,
    batch_size=32,
    shuffle=False,
    num_workers=2
)

print(train_ds.classes)
```

```python
import torch.nn as nn
from torchvision import models

device = "cuda" if torch.cuda.is_available() else "cpu"

model = models.efficientnet_b0(
    weights=models.EfficientNet_B0_Weights.DEFAULT
)

in_features = model.classifier[1].in_features

model.classifier[1] = nn.Linear(
    in_features,
    7
)

model = model.to(device)
```

```python
criterion = nn.CrossEntropyLoss()

optimizer = torch.optim.AdamW(
    model.parameters(),
    lr=1e-4
)

epochs = 25
```

```python
for epoch in range(epochs):

    model.train()

    running_loss = 0

    for images, labels in train_loader:

        images = images.to(device)
        labels = labels.to(device)

        optimizer.zero_grad()

        outputs = model(images)

        loss = criterion(outputs, labels)

        loss.backward()

        optimizer.step()

        running_loss += loss.item()

    print(
        f"Epoch {epoch+1}/{epochs} "
        f"Loss: {running_loss/len(train_loader):.4f}"
    )
```

```python
model.eval()

correct = 0
total = 0

with torch.no_grad():

    for images, labels in val_loader:

        images = images.to(device)
        labels = labels.to(device)

        outputs = model(images)

        preds = outputs.argmax(1)

        correct += (preds == labels).sum().item()
        total += labels.size(0)

print(
    f"Validation Accuracy: "
    f"{100*correct/total:.2f}%"
)
```

```python
from PIL import Image
from torchvision import transforms
import torch

classes = [
    "cut",
    "good",
    "husk",
    "immature",
    "parchment",
    "partial-black",
    "shell"
]

transform = transforms.Compose([
    transforms.Resize((224,224)),
    transforms.ToTensor()
])

img_path = "/content/drive/MyDrive/newMaskedDataset/newMaskedDataset/test/images/shell_75_jpg.rf.13717aae2a7e2cae2796bf112507f678.jpg"

img = Image.open(img_path).convert("RGB")

x = transform(img).unsqueeze(0).to(device)

with torch.no_grad():
    outputs = model(x)

pred = outputs.argmax(1).item()

print("Prediction:", classes[pred])
```
